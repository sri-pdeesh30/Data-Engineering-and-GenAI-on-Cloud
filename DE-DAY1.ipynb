```python
import pandas as pd

# Load dataset
df = pd.read_csv('student_performance.csv')

print("\nDATASET OVERVIEW")
print("-" * 30)

print("Total number of students :", len(df))
print("Total number of columns  :", len(df.columns))
print("Unique departments       :", df['department'].nunique())
```

```python
print("\nDEPARTMENT WISE ANALYSIS")
print("-" * 30)

dept_count = df['department'].value_counts()

print(dept_count)
```

```python
print("\nSCORE ANALYSIS")
print("-" * 30)

print("Highest Math Score            :", df['math_score'].max())
print("Lowest Math Score             :", df['math_score'].min())
print("Average Math Score            :", round(df['math_score'].mean(), 2))
print("Average Programming Score     :", round(df['programming_score'].mean(), 2))
print("Average Attendance Percentage :", round(df['attendance_percentage'].mean(), 2))
```

```python
# Creating total score column
df['total_score'] = (
    df['math_score'] +
    df['science_score'] +
    df['english_score'] +
    df['programming_score']
)

print("\nGENDER WISE ANALYSIS")
print("-" * 30)

gender_avg = df.groupby('gender')['total_score'].mean()

print(gender_avg)
```

```python
print("\nTOP PERFORMERS")
print("-" * 30)

top_students = df[['name', 'department', 'total_score']].sort_values(
    by='total_score',
    ascending=False
)

print(top_students.head(5))
```

```python
print("\nATTENDANCE ANALYSIS")
print("-" * 30)

attendance_data = df[['name', 'department', 'attendance_percentage']]

print(attendance_data)

if (df['attendance_percentage'] > 75).all():
    print("\nAll students have attendance above 75%")
else:
    print("\nSome students have attendance below 75%")
```
